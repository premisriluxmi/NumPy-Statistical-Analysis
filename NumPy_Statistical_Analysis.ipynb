# NumPy Statistical Analysis

## Objective

To analyze a numerical dataset using NumPy statistical functions, calculate important statistics, compare mean and median, and identify unusually high or low values.

```python
import numpy as np

marks = np.array([65, 70, 75, 80, 85, 90, 72, 78, 88, 95])

print("Student Marks:", marks)
print("Number of Observations:", len(marks))
```

```python
mean_value = np.mean(marks)
median_value = np.median(marks)
std_value = np.std(marks)
minimum_value = np.min(marks)
maximum_value = np.max(marks)

print("Mean:", round(mean_value, 2))
print("Median:", round(median_value, 2))
print("Standard Deviation:", round(std_value, 2))
print("Minimum:", minimum_value)
print("Maximum:", maximum_value)
```

```python
print("Mean:", round(mean_value, 2))
print("Median:", round(median_value, 2))

if mean_value > median_value:
    print("The mean is greater than the median.")
elif mean_value < median_value:
    print("The mean is less than the median.")
else:
    print("The mean and median are equal.")
```

```python
low_values = marks[marks < mean_value - std_value]
high_values = marks[marks > mean_value + std_value]

print("Unusually Low Values:", low_values)
print("Unusually High Values:", high_values)
```

```python
print("--------------------------------------")
print("     NumPy Statistical Analysis       ")
print("--------------------------------------")
print("Dataset:", marks)
print("Observations:", len(marks))
print("Mean:", round(mean_value, 2))
print("Median:", round(median_value, 2))
print("Standard Deviation:", round(std_value, 2))
print("Minimum:", minimum_value)
print("Maximum:", maximum_value)
print("Unusually Low Values:", low_values)
print("Unusually High Values:", high_values)
```

## Explanation of Results

A numerical dataset containing 10 student marks was created using NumPy.

The mean, median, standard deviation, minimum, and maximum were calculated using NumPy statistical functions.

The mean was 79.8, while the median was 79.0. The mean is slightly greater than the median.

The minimum mark was 65, and the maximum mark was 95. The standard deviation was approximately 9.12, indicating the spread of marks around the mean.

Values more than one standard deviation below or above the mean were identified for further inspection.

This task helped demonstrate how NumPy can be used to summarize numerical data and identify patterns in a dataset.
